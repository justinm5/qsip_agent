# Insider Analysis

Explore insider buying patterns and subsequent returns.

```python
from qsip.db import TimescaleDB
import polars as pl

db = TimescaleDB('postgres://qsip:qsip@localhost:5432/qsip?sslmode=disable')
with db.cursor() as cur:
    cur.execute("SELECT * FROM sec_filings WHERE form_type IN ('4','3','5') ORDER BY filed_at DESC LIMIT 1000")
    filings = cur.fetchall()
pl.DataFrame(filings).head()
```
