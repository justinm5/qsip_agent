# Factor Research

Explore raw factors: insider activity, sentiment, volume, and returns.

```python
import polars as pl
from qsip.db import TimescaleDB
from qsip.config import Config

cfg = Config.from_env()
db = TimescaleDB(cfg.db_dsn)
features = db.get_training_features(limit=10000)
df = pl.DataFrame([f[0] for f in features])
df.head()
```

```python
import matplotlib.pyplot as plt

df.plot.scatter(x='volume_ratio', y='return_20d', title='Volume Anomaly vs Forward Return')
```
